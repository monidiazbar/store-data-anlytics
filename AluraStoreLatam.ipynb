### Importación de datos



```python
import pandas as pd

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_4.csv"

tienda1 = pd.read_csv(url)
tienda2 = pd.read_csv(url2)
tienda3 = pd.read_csv(url3)
tienda4 = pd.read_csv(url4)

tienda1.head()
```

```python
tienda1.info()
```

```python
tiendas = [tienda1, tienda2, tienda3, tienda4]

for tienda in tiendas:
  print(tienda.info())
```

#1. Análisis de facturación



```python
resultados = []

todas_las_tiendas = [tienda1, tienda2, tienda3, tienda4]

for i, df in enumerate(todas_las_tiendas):
    tienda_num = i + 1  # Get the store number (1-indexed)

    facturacion = df['Precio'].sum()

    resultados.append({
        'Tienda': tienda_num,
        'Facturacion_Total': facturacion,
        'Ventas_Registradas': len(df),
        'Precio_Promedio': df['Precio'].mean()
    })

df_resultados = pd.DataFrame(resultados)
df_resultados = df_resultados.sort_values('Facturacion_Total')
```

```python
pd.options.display.float_format = '{:,.2f}'.format
df_resultados
```

```python
# Ventas por Forma de pago

formas_pago = []

for i, df in enumerate(todas_las_tiendas):
  tienda_num = i + 1
  df_formas_pago_tienda = df.groupby('Método de pago')['Precio'].sum().reset_index()
  df_formas_pago_tienda['Tienda'] = tienda_num

  formas_pago.append(df_formas_pago_tienda)

df_formas_pago = pd.concat(formas_pago)
df_formas_pago = df_formas_pago.rename(columns={'Precio': 'Total ventas'})

df_formas_pago_por_tienda = df_formas_pago.pivot(index= 'Método de pago', columns= ['Tienda'], values= 'Total ventas')
df_formas_pago_por_tienda.columns = [f'Tienda_{col}' for col in df_formas_pago_por_tienda.columns]
df_formas_pago_por_tienda.reset_index(inplace=True)

# Calcular las ventas totales globales por método de pago
total_ventas_globales = df_formas_pago.groupby('Método de pago')['Total ventas'].sum().reset_index()
total_ventas_globales = total_ventas_globales.rename(columns={'Total ventas': 'Ventas_Totales_Globales'})

# Combinar la tabla pivote con las ventas totales globales
df_formas_pago_final = pd.merge(df_formas_pago_por_tienda, total_ventas_globales, on='Método de pago', how='left')

df_formas_pago_final.sort_values(by='Ventas_Totales_Globales', ascending=False)
```

# 2. Ventas por categoría

```python
ventas_por_categoria = []

for i, df in enumerate(todas_las_tiendas):
    tienda_num = i + 1
    ventas_categoria_tienda = df.groupby('Categoría del Producto')['Precio'].sum().reset_index()
    ventas_categoria_tienda['Tienda'] = tienda_num
    ventas_por_categoria.append(ventas_categoria_tienda)

df_ventas_por_categoria = pd.concat(ventas_por_categoria)

# Agrupar por categoría a través de todas las tiendas para obtener un resumen general
df_ventas_categoria_total = df_ventas_por_categoria.groupby('Categoría del Producto')['Precio'].sum().reset_index()
df_ventas_categoria_total = df_ventas_categoria_total.rename(columns={'Precio': 'Ventas_Totales'})
df_ventas_categoria_total = df_ventas_categoria_total.sort_values(by='Ventas_Totales', ascending=False)

#Agrupar Categorías por tienda
df_ventas_categoria_tienda = df_ventas_por_categoria.groupby(['Tienda', 'Categoría del Producto'])['Precio'].sum().reset_index()
df_ventas_categoria_tienda = df_ventas_categoria_tienda.rename(columns={'Precio': 'Ventas_por_Tienda'})

df_ventas_pivot = df_ventas_categoria_tienda.pivot(index='Categoría del Producto', columns='Tienda', values='Ventas_por_Tienda')
df_ventas_categoria_final = pd.merge(df_ventas_categoria_total, df_ventas_pivot, on='Categoría del Producto', how='left')
# Renombrar las columnas de tienda para mayor claridad, por ejemplo 'Tienda_1', 'Tienda_2', etc.
df_ventas_categoria_final.columns = ['Categoría del Producto', 'Ventas_Totales'] + [f'Tienda_{col}' for col in
                                                                                    df_ventas_categoria_final.columns[2:]]

df_ventas_categoria_final
```

# 3. Calificación promedio de la tienda


```python
calificaciones_promedio = []

for i, df in enumerate(todas_las_tiendas):
    tienda_num = i + 1
    calificacion_promedio = df['Calificación'].mean()
    calificaciones_promedio.append({'Tienda': tienda_num, 'Calificacion_Promedio': calificacion_promedio})

df_calificaciones_promedio = pd.DataFrame(calificaciones_promedio)
df_calificaciones_promedio = df_calificaciones_promedio.sort_values(by='Calificacion_Promedio', ascending=False)
```

# 4. Productos más y menos vendidos

```python
top_bottom_productos_por_tienda = []

for i, df in enumerate(todas_las_tiendas):
    tienda_num = i + 1
    ventas_por_producto = df.groupby('Producto')['Precio'].sum().reset_index()
    ventas_por_producto = ventas_por_producto.rename(columns={'Precio': 'Ventas_Totales'})

    # Productos más vendidos por tienda
    top_productos = ventas_por_producto.nlargest(5, 'Ventas_Totales')
    top_productos['Tipo'] = 'Top 5'
    top_productos['Tienda'] = tienda_num

    # Productos menos vendidos por tienda
    bottom_productos = ventas_por_producto.nsmallest(5, 'Ventas_Totales')
    bottom_productos['Tipo'] = 'Bottom 5'
    bottom_productos['Tienda'] = tienda_num

    top_bottom_productos_por_tienda.append(top_productos)
    top_bottom_productos_por_tienda.append(bottom_productos)

df_top_bottom_productos_por_tienda = pd.concat(top_bottom_productos_por_tienda)

print('Top 5 y Bottom 5 Productos por Tienda:')
print(df_top_bottom_productos_por_tienda)
```

```python
# Análisis global de productos más y menos vendidos
ventas_por_producto_global = pd.concat(todas_las_tiendas).groupby('Producto')['Precio'].sum().reset_index()
ventas_por_producto_global = ventas_por_producto_global.rename(columns={'Precio': 'Ventas_Totales'})

# Top 5 productos global
top_productos_global = ventas_por_producto_global.nlargest(5, 'Ventas_Totales')
top_productos_global['Tipo'] = 'Top 5 Global'

# Bottom 5 productos global
bottom_productos_global = ventas_por_producto_global.nsmallest(5, 'Ventas_Totales')
bottom_productos_global['Tipo'] = 'Bottom 5 Global'

df_productos_global = pd.concat([top_productos_global, bottom_productos_global])

print('\nTop 5 y Bottom 5 Productos Globales:')
print(df_productos_global)
```

# 5. Envío promedio por tienda

```python
envios_promedio = []

for i, df in enumerate(todas_las_tiendas):
    tienda_num = i + 1
    envio_promedio = df['Costo de envío'].mean()
    envios_promedio.append({'Tienda': tienda_num, 'Envio_Promedio': envio_promedio})

df_envios_promedio = pd.DataFrame(envios_promedio)
df_envios_promedio = df_envios_promedio.sort_values(by='Envio_Promedio', ascending=False)
```
